```python
import pandas as pd
df=pd.read_csv(r"/content/employee_attendance_dataset.csv")
print(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))
sns.histplot(df["attendence percentage"], bins=10, kde=True)
plt.title("Attendance Percentage Distribution")
plt.xlabel("Attendance Percentage")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv(r"/content/employee_attendance_dataset.csv")
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="overtime", y="attendence percentage", hue="Department")
plt.title("Overtime Hours vs. Attendance Percentage")
plt.xlabel("Overtime Hours")
plt.ylabel("Attendance Percentage (%)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 5))
sns.lineplot(data=df, x="Employee_ID", y="attendence percentage", marker="o", color="crimson")
plt.title("Attendance Percentage Outlier Detection by Employee")
plt.xlabel("Employee ID")
plt.ylabel("Attendance Percentage (%)")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
gender_attendance = df.groupby('gender')['attendence percentage'].mean()
plt.figure(figsize=(6, 6))
plt.pie(gender_attendance, labels=gender_attendance.index, autopct='%1.1f%%', colors=['skyblue', 'pink'], startangle=90)
plt.title("Average Attendance Percentage Distribution by Gender")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="gender", y="attendence percentage", ci=None, palette="pastel")
plt.title("Average Attendance Percentage by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Attendance Percentage (%)")
plt.tight_layout()
plt.show()
```
